```python
from pathlib import Path
from collections import Counter
import random
import json

DATASET = Path("/content/ViolenceDetectionImproved/safe_dataset")
EXPERIMENT = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/"
    "experiments/improved_v1"
)
MANIFEST = EXPERIMENT / "split_manifest.json"

EXTENSIONS = {".avi", ".mp4", ".mov", ".mkv", ".webm"}
SEED = 42

assert DATASET.is_dir(), "Local dataset missing. Rerun the copy cell."

EXPERIMENT.mkdir(parents=True, exist_ok=True)

if MANIFEST.exists():
    split_manifest = json.loads(MANIFEST.read_text())
    print("Reusing the saved split.")
else:
    split_manifest = {
        "seed": SEED,
        "description": (
            "Sorted filenames, then seeded per-class shuffle; "
            "90/10 split of official train. Official val held out."
        ),
        "train": [],
        "validation": [],
        "test": [],
    }

    for label, class_name in [(0, "NonFight"), (1, "Fight")]:
        files = sorted(
            (
                p for p in (DATASET / "train" / class_name).rglob("*")
                if p.is_file() and p.suffix.lower() in EXTENSIONS
            ),
            key=lambda p: p.relative_to(DATASET).as_posix(),
        )
        assert len(files) == 800, f"Unexpected count for {class_name}"

        random.Random(SEED + label).shuffle(files)
        validation_count = 80

        for split_name, selected in [
            ("validation", files[:validation_count]),
            ("train", files[validation_count:]),
        ]:
            split_manifest[split_name].extend(
                {
                    "path": p.relative_to(DATASET).as_posix(),
                    "label": label,
                }
                for p in selected
            )

        test_files = sorted(
            (
                p for p in (DATASET / "val" / class_name).rglob("*")
                if p.is_file() and p.suffix.lower() in EXTENSIONS
            ),
            key=lambda p: p.relative_to(DATASET).as_posix(),
        )
        assert len(test_files) == 200, f"Unexpected test count: {class_name}"

        split_manifest["test"].extend(
            {
                "path": p.relative_to(DATASET).as_posix(),
                "label": label,
            }
            for p in test_files
        )

    for split_name in ("train", "validation"):
        random.Random(SEED).shuffle(split_manifest[split_name])

    # Exclusive creation prevents overwriting an existing manifest.
    with MANIFEST.open("x", encoding="utf-8") as f:
        json.dump(split_manifest, f, indent=2)

    print("Created and saved a new fixed split.")

expected_sizes = {"train": 1440, "validation": 160, "test": 400}
path_sets = {}

for split_name, expected in expected_sizes.items():
    records = split_manifest[split_name]
    paths = [record["path"] for record in records]

    assert len(records) == expected
    assert len(set(paths)) == expected, "Duplicate paths found."
    assert all((DATASET / p).is_file() for p in paths), "Missing video files."

    path_sets[split_name] = set(paths)
    counts = Counter(record["label"] for record in records)
    assert counts[0] == counts[1] == expected // 2

    print(
        f"{split_name}: {len(records)} clips "
        f"(NonFight={counts[0]}, Fight={counts[1]})"
    )

assert not (path_sets["train"] & path_sets["validation"])
assert not (path_sets["train"] & path_sets["test"])
assert not (path_sets["validation"] & path_sets["test"])

train_items = [
    (DATASET / r["path"], r["label"])
    for r in split_manifest["train"]
]
validation_items = [
    (DATASET / r["path"], r["label"])
    for r in split_manifest["validation"]
]

print("\nNo file-path overlap between splits.")
print("Manifest saved at:", MANIFEST)
```

```python
from pathlib import Path
import json
import cv2
from tqdm.auto import tqdm

DATASET = Path("/content/ViolenceDetectionImproved/safe_dataset")
EXPERIMENT = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/"
    "experiments/improved_v1"
)

manifest = json.loads(
    (EXPERIMENT / "split_manifest.json").read_text()
)

report = []
problems = []

for split_name in ("train", "validation"):
    for record in tqdm(manifest[split_name], desc=f"Checking {split_name}"):
        path = DATASET / record["path"]
        capture = cv2.VideoCapture(str(path))

        opened = capture.isOpened()
        expected_frames = (
            int(capture.get(cv2.CAP_PROP_FRAME_COUNT)) if opened else 0
        )
        decoded_frames = 0

        try:
            while opened:
                ok, frame = capture.read()
                if not ok:
                    break
                if frame is None or frame.size == 0:
                    break
                decoded_frames += 1
        finally:
            capture.release()

        issues = []
        if not opened:
            issues.append("could not open")
        elif decoded_frames == 0:
            issues.append("no decodable frames")
        else:
            if decoded_frames < 16:
                issues.append("fewer than 16 frames")
            if expected_frames > 0 and decoded_frames < expected_frames:
                issues.append("decoded fewer frames than metadata reports")

        result = {
            "split": split_name,
            "path": record["path"],
            "reported_frames": expected_frames,
            "decoded_frames": decoded_frames,
            "issues": issues,
        }
        report.append(result)

        if issues:
            problems.append(result)

report_path = EXPERIMENT / "video_decode_report.json"
report_path.write_text(json.dumps(report, indent=2))

print(f"\nChecked {len(report)} videos.")
print(f"Videos needing review: {len(problems)}")

for result in problems[:20]:
    print(result["path"], ":", "; ".join(result["issues"]))

if len(problems) > 20:
    print("Additional entries are in the saved report.")

print("\nReport saved at:", report_path)
```

```python
from pathlib import Path
import json
import random
import cv2
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DATASET = Path("/content/ViolenceDetectionImproved/safe_dataset")
EXPERIMENT = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/"
    "experiments/improved_v1"
)

manifest = json.loads(
    (EXPERIMENT / "split_manifest.json").read_text()
)

class ImprovedVideoDataset(Dataset):
    def __init__(self, records, training=False):
        self.records = records
        self.training = training
        self.sequence_length = 16
        self.image_size = 160
        self.mean = np.array([0.485, 0.456, 0.406], dtype=np.float32)
        self.std = np.array([0.229, 0.224, 0.225], dtype=np.float32)

    def __len__(self):
        return len(self.records)

    def __getitem__(self, index):
        record = self.records[index]
        path = DATASET / record["path"]
        cap = cv2.VideoCapture(str(path))

        try:
            total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
            if not cap.isOpened() or total < self.sequence_length:
                raise RuntimeError(f"Cannot sample 16 frames: {path}")

            if self.training:
                # Choose one frame from each temporal segment.
                edges = np.linspace(
                    0, total, self.sequence_length + 1, dtype=int
                )
                indices = [
                    random.randrange(int(edges[i]), int(edges[i + 1]))
                    for i in range(self.sequence_length)
                ]

                # One set of augmentation parameters for the whole clip.
                flip = random.random() < 0.5
                brightness = random.uniform(-0.05, 0.05)
                contrast = random.uniform(0.9, 1.1)
            else:
                # Match the original deterministic validation sampling.
                indices = np.linspace(
                    0, total - 1, self.sequence_length
                ).astype(int)
                flip = False
                brightness = 0.0
                contrast = 1.0

            frames = []
            for frame_index in indices:
                cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_index))
                ok, frame = cap.read()
                if not ok or frame is None:
                    raise RuntimeError(
                        f"Failed to read frame {frame_index}: {path}"
                    )

                frame = cv2.resize(
                    frame, (self.image_size, self.image_size)
                )
                frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                frame = frame.astype(np.float32) / 255.0

                if flip:
                    frame = frame[:, ::-1, :]

                frame = np.clip(
                    (frame - 0.5) * contrast + 0.5 + brightness,
                    0.0,
                    1.0,
                )
                frame = (frame - self.mean) / self.std
                frames.append(np.transpose(frame, (2, 0, 1)))

            clip = torch.from_numpy(np.stack(frames)).float()
            label = torch.tensor(record["label"], dtype=torch.float32)
            return clip, label
        finally:
            cap.release()

train_dataset = ImprovedVideoDataset(manifest["train"], training=True)
validation_dataset = ImprovedVideoDataset(
    manifest["validation"], training=False
)

loader_generator = torch.Generator().manual_seed(SEED)

train_loader = DataLoader(
    train_dataset,
    batch_size=4,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
    generator=loader_generator,
)

validation_loader = DataLoader(
    validation_dataset,
    batch_size=4,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
)

# Smoke-check dimensions, finite values, and deterministic validation.
clips, labels = next(iter(train_loader))
assert clips.shape == (4, 16, 3, 160, 160)
assert torch.isfinite(clips).all()

validation_clip_1, _ = validation_dataset[0]
validation_clip_2, _ = validation_dataset[0]
assert torch.equal(validation_clip_1, validation_clip_2)

# Reset random states after the checks, before training.
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
loader_generator.manual_seed(SEED)

print("Training batches:", len(train_loader))
print("Validation batches:", len(validation_loader))
print("Clip shape:", tuple(clips.shape))
print("Labels:", labels.tolist())
print("All sampled values are finite.")
print("Validation sampling is deterministic.")
print("Data loaders ready. Training has not started.")
```

```python
import sys
import random
import json
import os
import numpy as np
import torch
from pathlib import Path
from torch import nn
from tqdm.auto import tqdm

# ------------------------------------------------------------
# Paths and prerequisites
# ------------------------------------------------------------

BACKEND = Path("/content/ViolenceDetectionImproved/backend")
EXPERIMENT = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/"
    "experiments/improved_v1"
)

assert BACKEND.is_dir(), "Local backend missing. Rerun the copy cell."
assert EXPERIMENT.is_dir(), "Experiment folder missing. Check Drive mount."
assert torch.cuda.is_available(), "Enable the GPU before training."
assert all(
    name in globals()
    for name in (
        "train_loader",
        "validation_loader",
        "loader_generator",
    )
), "Run the data-loader cell first."

manifest = json.loads(
    (EXPERIMENT / "split_manifest.json").read_text()
)

sys.path.insert(0, str(BACKEND))

from action_model import ViolenceCNNLSTM
from train_violence_model import (
    empty_metrics,
    update_metrics,
    finalise_metrics,
    metrics_text,
)

BEST_PATH = EXPERIMENT / "stage1_best.pt"
LATEST_PATH = EXPERIMENT / "stage1_latest.pt"

# Prevent an accidental restart from replacing saved progress.
assert not BEST_PATH.exists() and not LATEST_PATH.exists(), (
    "Stage-1 checkpoints already exist. Stop here so we can "
    "resume the saved run instead of overwriting it."
)

# ------------------------------------------------------------
# Reproducibility and settings
# ------------------------------------------------------------

DEVICE = torch.device("cuda")
SEED = 42
MAX_EPOCHS = 10
EARLY_STOPPING_PATIENCE = 5

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
loader_generator.manual_seed(SEED)

torch.backends.cudnn.benchmark = False
torch.backends.cudnn.deterministic = True

# ------------------------------------------------------------
# Fresh pretrained model
# ------------------------------------------------------------

model = ViolenceCNNLSTM(
    hidden_size=64,
    lstm_layers=1,
    pretrained=True,
).to(DEVICE)

# Freeze MobileNet weights.
for parameter in model.cnn.parameters():
    parameter.requires_grad = False

trainable_parameters = [
    parameter
    for parameter in model.parameters()
    if parameter.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_parameters,
    lr=1e-4,
    weight_decay=1e-4,
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
)

loss_fn = nn.BCEWithLogitsLoss()

config = {
    "sequence_length": 16,
    "image_size": 160,
    "hidden_size": 64,
    "lstm_layers": 1,
    "threshold": 0.80,
    "backbone_frozen": True,
}

history = []
best_f1 = -1.0
best_epoch = 0
epochs_without_improvement = 0

# ------------------------------------------------------------
# Checkpoint saving
# ------------------------------------------------------------

def save_checkpoint(payload, destination):
    temporary = destination.with_suffix(".tmp")
    torch.save(payload, temporary)
    os.replace(temporary, destination)


# ------------------------------------------------------------
# Training / validation function
# ------------------------------------------------------------

def run_epoch(loader, training, description):
    if training:
        model.train()

        # Keep frozen CNN BatchNorm statistics unchanged.
        model.cnn.eval()
    else:
        model.eval()

    counters = empty_metrics()
    progress = tqdm(loader, desc=description, unit="batch")

    with torch.set_grad_enabled(training):
        for clips, labels in progress:
            clips = clips.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            if training:
                optimizer.zero_grad(set_to_none=True)

            logits = model(clips)
            loss = loss_fn(logits, labels)

            if not torch.isfinite(loss):
                raise RuntimeError(
                    "Non-finite loss detected. Training stopped."
                )

            if training:
                loss.backward()

                nn.utils.clip_grad_norm_(
                    trainable_parameters,
                    max_norm=1.0,
                    error_if_nonfinite=True,
                )

                optimizer.step()

            update_metrics(counters, logits, labels, loss)
            current = finalise_metrics(counters)

            progress.set_postfix(
                loss=f"{current['loss']:.4f}",
                accuracy=f"{current['accuracy']:.1%}",
                precision=f"{current['precision']:.1%}",
                recall=f"{current['recall']:.1%}",
                f1=f"{current['f1']:.1%}",
            )

    return finalise_metrics(counters)


# ------------------------------------------------------------
# Train for at most 10 epochs
# ------------------------------------------------------------

print("GPU:", torch.cuda.get_device_name(0))
print("Maximum epochs:", MAX_EPOCHS)
print("Early-stopping patience:", EARLY_STOPPING_PATIENCE)
print("Training clips:", len(train_loader.dataset))
print("Validation clips:", len(validation_loader.dataset))
print(
    "Trainable parameters:",
    f"{sum(p.numel() for p in trainable_parameters):,}",
)
print("MobileNet weights and BatchNorm statistics are frozen.")
print("Starting a fresh stage-1 experiment.\n")

for epoch in range(1, MAX_EPOCHS + 1):
    learning_rate_used = optimizer.param_groups[0]["lr"]

    training_metrics = run_epoch(
        train_loader,
        training=True,
        description=f"Training {epoch}/{MAX_EPOCHS}",
    )

    validation_metrics = run_epoch(
        validation_loader,
        training=False,
        description=f"Validation {epoch}/{MAX_EPOCHS}",
    )

    # Reduce LR when validation loss stalls.
    scheduler.step(validation_metrics["loss"])
    next_learning_rate = optimizer.param_groups[0]["lr"]

    # Select checkpoints by validation F1 at probability threshold 0.5.
    # The dashboard alert threshold of 0.80 is a separate setting.
    improved = validation_metrics["f1"] > best_f1

    if improved:
        best_f1 = validation_metrics["f1"]
        best_epoch = epoch
        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1

    history.append({
        "epoch": epoch,
        "training": training_metrics,
        "validation": validation_metrics,
        "learning_rate": learning_rate_used,
        "next_learning_rate": next_learning_rate,
    })

    # Save state needed to recover after a runtime interruption.
    checkpoint = {
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
        "config": config,
        "training_settings": {
            "seed": SEED,
            "max_epochs": MAX_EPOCHS,
            "early_stopping_patience": EARLY_STOPPING_PATIENCE,
            "gradient_clip_norm": 1.0,
            "metric_probability_threshold": 0.5,
        },
        "epoch": epoch,
        "best_epoch": best_epoch,
        "best_validation_f1": best_f1,
        "epochs_without_improvement": epochs_without_improvement,
        "training_metrics": training_metrics,
        "validation_metrics": validation_metrics,
        "history": history,
        "split_manifest": manifest,
        "rng_state": {
            "python": random.getstate(),
            "numpy": np.random.get_state(),
            "torch": torch.get_rng_state(),
            "cuda": torch.cuda.get_rng_state_all(),
            "loader": loader_generator.get_state(),
        },
    }

    if improved:
        save_checkpoint(checkpoint, BEST_PATH)

    save_checkpoint(checkpoint, LATEST_PATH)

    temporary_history = EXPERIMENT / "stage1_history.tmp"
    temporary_history.write_text(
        json.dumps(history, indent=2),
        encoding="utf-8",
    )
    os.replace(
        temporary_history,
        EXPERIMENT / "stage1_history.json",
    )

    print("\n" + "=" * 72)
    print(f"EPOCH {epoch}/{MAX_EPOCHS}")
    print("=" * 72)
    print(metrics_text("Training", training_metrics))
    print(metrics_text("Validation", validation_metrics))
    print(
        f"Learning rate: {learning_rate_used:.2g} "
        f"→ {next_learning_rate:.2g}"
    )
    print(
        f"Best epoch: {best_epoch} | "
        f"Best validation F1: {best_f1:.1%}"
    )
    print(
        "No-improvement count: "
        f"{epochs_without_improvement}/{EARLY_STOPPING_PATIENCE}"
    )

    if improved:
        print("New best checkpoint saved.")

    print("Latest checkpoint and history saved to Drive.\n")

    if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:
        print("Early stopping: validation F1 has not improved.")
        break

# ------------------------------------------------------------
# Final summary
# ------------------------------------------------------------

best_record = next(
    record for record in history
    if record["epoch"] == best_epoch
)

print("\n" + "=" * 72)
print("STAGE-1 RESULTS")
print("=" * 72)
print("Completed epochs:", history[-1]["epoch"])
print("Selected best epoch:", best_epoch)
print(
    metrics_text(
        "Training at selected epoch",
        best_record["training"],
    )
)
print(
    metrics_text(
        "Validation at selected epoch",
        best_record["validation"],
    )
)
print("\nBest checkpoint:", BEST_PATH)
print("Latest checkpoint:", LATEST_PATH)
print("History:", EXPERIMENT / "stage1_history.json")
print("\nTest set was not evaluated.")
```

```python
from pathlib import Path
import sys
import json
import random
import numpy as np
import torch
from torch import nn

BACKEND = Path("/content/ViolenceDetectionImproved/backend")
SOURCE_EXPERIMENT = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/"
    "experiments/lstm32_v1"
)
FINETUNE_EXPERIMENT = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/"
    "experiments/lstm32_finetune_v1"
)
SOURCE_CHECKPOINT = SOURCE_EXPERIMENT / "stage1_best.pt"

assert torch.cuda.is_available(), "Enable the GPU first."
assert SOURCE_CHECKPOINT.is_file(), "Best checkpoint is missing."
assert all(
    name in globals()
    for name in ("train_loader", "validation_loader", "loader_generator")
), "Data loaders are missing. Tell me if the Colab runtime restarted."

FINETUNE_EXPERIMENT.mkdir(parents=True, exist_ok=True)

assert not any(FINETUNE_EXPERIMENT.glob("*.pt")), (
    "Fine-tuning checkpoints already exist. Stop here to avoid "
    "replacing saved progress."
)

sys.path.insert(0, str(BACKEND))
from action_model import ViolenceCNNLSTM

# Load your own saved checkpoint onto CPU first.
source_checkpoint = torch.load(
    SOURCE_CHECKPOINT,
    map_location="cpu",
    weights_only=False,
)

finetune_config = dict(source_checkpoint["config"])
finetune_manifest = source_checkpoint["split_manifest"]

# Verify the same training/validation membership and preprocessing.
for split_name, loader, training in [
    ("train", train_loader, True),
    ("validation", validation_loader, False),
]:
    assert loader.dataset.records == finetune_manifest[split_name]
    assert loader.dataset.training == training
    assert (
        loader.dataset.sequence_length
        == finetune_config["sequence_length"]
    )
    assert loader.dataset.image_size == finetune_config["image_size"]
    assert loader.batch_size == 4
    assert loader.num_workers == 0

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
loader_generator.manual_seed(SEED)

torch.backends.cudnn.benchmark = False
torch.backends.cudnn.deterministic = True

# Construct the exact architecture stored in the checkpoint.
# No new ImageNet download is needed: checkpoint includes CNN weights.
finetune_model = ViolenceCNNLSTM(
    hidden_size=finetune_config["hidden_size"],
    lstm_layers=finetune_config["lstm_layers"],
    pretrained=False,
)
finetune_model.load_state_dict(source_checkpoint["model_state"])

# Freeze the whole CNN first.
for parameter in finetune_model.cnn.parameters():
    parameter.requires_grad = False

# Unfreeze only the final two top-level MobileNet feature blocks.
block_names = list(finetune_model.cnn._modules.keys())
unfrozen_blocks = block_names[-2:]

for name in unfrozen_blocks:
    for parameter in finetune_model.cnn._modules[name].parameters():
        parameter.requires_grad = True

# Freeze both BatchNorm parameters and running statistics.
def freeze_batchnorm(module):
    for layer in module.modules():
        if isinstance(layer, nn.modules.batchnorm._BatchNorm):
            layer.eval()
            for parameter in layer.parameters():
                parameter.requires_grad = False

def set_finetune_training_mode():
    finetune_model.train()

    # Earlier frozen blocks stay in evaluation mode.
    finetune_model.cnn.eval()

    # Only selected blocks enter training mode.
    for name in unfrozen_blocks:
        finetune_model.cnn._modules[name].train()

    # Override BatchNorm inside the selected blocks too.
    freeze_batchnorm(finetune_model)

freeze_batchnorm(finetune_model)

DEVICE = torch.device("cuda")
finetune_model = finetune_model.to(DEVICE)

cnn_parameters = [
    parameter
    for parameter in finetune_model.cnn.parameters()
    if parameter.requires_grad
]
head_parameters = [
    parameter
    for component in (finetune_model.lstm, finetune_model.classifier)
    for parameter in component.parameters()
    if parameter.requires_grad
]

finetune_optimizer = torch.optim.AdamW(
    [
        {"params": cnn_parameters, "lr": 1e-5},
        {"params": head_parameters, "lr": 5e-5},
    ],
    weight_decay=1e-4,
)

finetune_scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    finetune_optimizer,
    mode="min",
    factor=0.5,
    patience=1,
    min_lr=[1e-7, 1e-6],
)

set_finetune_training_mode()

# Verify BatchNorm really is frozen.
batchnorm_layers = [
    layer
    for layer in finetune_model.modules()
    if isinstance(layer, nn.modules.batchnorm._BatchNorm)
]
assert all(not layer.training for layer in batchnorm_layers)
assert all(
    not parameter.requires_grad
    for layer in batchnorm_layers
    for parameter in layer.parameters()
)

finetune_config["backbone_frozen"] = False
finetune_config["unfrozen_cnn_blocks"] = unfrozen_blocks
finetune_config["batchnorm_frozen"] = True

manifest_path = FINETUNE_EXPERIMENT / "split_manifest.json"
if manifest_path.exists():
    assert json.loads(manifest_path.read_text()) == finetune_manifest
else:
    manifest_path.write_text(json.dumps(finetune_manifest, indent=2))

print("GPU:", torch.cuda.get_device_name(0))
print("Starting checkpoint:", SOURCE_CHECKPOINT)
print("Starting checkpoint epoch:", source_checkpoint["epoch"])
print(
    "Starting validation F1:",
    f"{source_checkpoint['validation_metrics']['f1']:.2%}",
)
print("LSTM hidden size:", finetune_config["hidden_size"])
print("Unfrozen MobileNet blocks:", unfrozen_blocks)
print(
    "Trainable CNN parameters:",
    f"{sum(p.numel() for p in cnn_parameters):,}",
)
print(
    "Trainable LSTM/classifier parameters:",
    f"{sum(p.numel() for p in head_parameters):,}",
)
print("CNN learning rate: 0.00001")
print("LSTM/classifier learning rate: 0.00005")
print("All BatchNorm layers frozen: YES")
print("Output folder:", FINETUNE_EXPERIMENT)
print("\nPreparation complete. Training has NOT started.")
```

```python
import os
import json
import random
import numpy as np
import torch
from torch import nn
from tqdm.auto import tqdm

from train_violence_model import (
    empty_metrics,
    update_metrics,
    finalise_metrics,
    metrics_text,
)

# ------------------------------------------------------------
# Settings and output files
# ------------------------------------------------------------

MAX_FINETUNE_EPOCHS = 10
PATIENCE = 5

FT_BEST = FINETUNE_EXPERIMENT / "finetune_best.pt"
FT_LATEST = FINETUNE_EXPERIMENT / "finetune_latest.pt"
FT_HISTORY = FINETUNE_EXPERIMENT / "finetune_history.json"

assert not FT_BEST.exists() and not FT_LATEST.exists(), (
    "Fine-tuning checkpoints already exist. Stop here so we "
    "can resume rather than overwrite them."
)

loss_fn = nn.BCEWithLogitsLoss()
ft_parameters = cnn_parameters + head_parameters

ft_history = []
ft_best_f1 = float(source_checkpoint["validation_metrics"]["f1"])
ft_best_epoch = 0  # 0 means the original starting checkpoint.
ft_no_improvement = 0
ft_epochs_at_reduced_lr = 0
initial_lrs = [
    group["lr"] for group in finetune_optimizer.param_groups
]


def ft_save(payload, destination):
    temporary = destination.with_suffix(".tmp")
    torch.save(payload, temporary)
    os.replace(temporary, destination)


def ft_run_epoch(loader, training, description):
    if training:
        set_finetune_training_mode()
    else:
        finetune_model.eval()

    counters = empty_metrics()
    progress = tqdm(loader, desc=description, unit="batch")

    with torch.set_grad_enabled(training):
        for clips, labels in progress:
            clips = clips.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            if training:
                finetune_optimizer.zero_grad(set_to_none=True)

            logits = finetune_model(clips)
            loss = loss_fn(logits, labels)

            if not torch.isfinite(loss):
                raise RuntimeError("Non-finite loss. Training stopped.")

            if training:
                loss.backward()
                nn.utils.clip_grad_norm_(
                    ft_parameters,
                    max_norm=1.0,
                    error_if_nonfinite=True,
                )
                finetune_optimizer.step()

            update_metrics(counters, logits, labels, loss)
            current = finalise_metrics(counters)

            progress.set_postfix(
                loss=f"{current['loss']:.4f}",
                accuracy=f"{current['accuracy']:.1%}",
                precision=f"{current['precision']:.1%}",
                recall=f"{current['recall']:.1%}",
                f1=f"{current['f1']:.1%}",
            )

    return finalise_metrics(counters)


# ------------------------------------------------------------
# Recheck the starting model before any weight updates
# ------------------------------------------------------------

starting_validation = ft_run_epoch(
    validation_loader,
    training=False,
    description="Checking starting checkpoint",
)

print(metrics_text("Starting validation", starting_validation))

assert abs(starting_validation["f1"] - ft_best_f1) < 1e-6, (
    "Starting validation F1 differs from the saved result. "
    "Stop and check the split and preprocessing before training."
)

# Initialize the scheduler against the starting validation loss.
finetune_scheduler.step(starting_validation["loss"])

# Save the starting model as the fallback best checkpoint.
ft_save(
    {
        "model_state": finetune_model.state_dict(),
        "config": finetune_config,
        "epoch": 0,
        "stage": "limited_finetuning",
        "source_checkpoint": str(SOURCE_CHECKPOINT),
        "validation_metrics": starting_validation,
        "best_validation_f1": ft_best_f1,
        "split_manifest": finetune_manifest,
    },
    FT_BEST,
)

# Reset random states so the preliminary validation check
# does not change the training sampling sequence.
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
loader_generator.manual_seed(SEED)

print(f"\nStarting fine-tuning. F1 to beat: {ft_best_f1:.2%}\n")

# ------------------------------------------------------------
# Fine-tuning loop
# ------------------------------------------------------------

for epoch in range(1, MAX_FINETUNE_EPOCHS + 1):
    lrs_used = [
        group["lr"] for group in finetune_optimizer.param_groups
    ]

    training_metrics = ft_run_epoch(
        train_loader, True,
        f"Fine-tune {epoch}/{MAX_FINETUNE_EPOCHS}",
    )
    validation_metrics = ft_run_epoch(
        validation_loader, False,
        f"Validation {epoch}/{MAX_FINETUNE_EPOCHS}",
    )

    # Count only epochs actually trained at a reduced rate.
    if any(
        used < initial
        for used, initial in zip(lrs_used, initial_lrs)
    ):
        ft_epochs_at_reduced_lr += 1

    finetune_scheduler.step(validation_metrics["loss"])
    next_lrs = [
        group["lr"] for group in finetune_optimizer.param_groups
    ]

    improved = validation_metrics["f1"] > ft_best_f1

    if improved:
        ft_best_f1 = validation_metrics["f1"]
        ft_best_epoch = epoch
        ft_no_improvement = 0
    else:
        ft_no_improvement += 1

    ft_history.append({
        "epoch": epoch,
        "training": training_metrics,
        "validation": validation_metrics,
        "learning_rates": lrs_used,
        "next_learning_rates": next_lrs,
    })

    payload = {
        "model_state": finetune_model.state_dict(),
        "optimizer_state": finetune_optimizer.state_dict(),
        "scheduler_state": finetune_scheduler.state_dict(),
        "config": finetune_config,
        "stage": "limited_finetuning",
        "source_checkpoint": str(SOURCE_CHECKPOINT),
        "epoch": epoch,
        "best_epoch": ft_best_epoch,
        "best_validation_f1": ft_best_f1,
        "epochs_without_improvement": ft_no_improvement,
        "epochs_at_reduced_lr": ft_epochs_at_reduced_lr,
        "initial_learning_rates": initial_lrs,
        "starting_validation_metrics": starting_validation,
        "training_settings": {
            "max_epochs": MAX_FINETUNE_EPOCHS,
            "patience": PATIENCE,
            "minimum_epochs_at_reduced_lr_before_early_stop": 2,
            "metric_probability_threshold": 0.5,
            "seed": SEED,
        },
        "training_metrics": training_metrics,
        "validation_metrics": validation_metrics,
        "history": ft_history,
        "split_manifest": finetune_manifest,
        "rng_state": {
            "python": random.getstate(),
            "numpy": np.random.get_state(),
            "torch": torch.get_rng_state(),
            "cuda": torch.cuda.get_rng_state_all(),
            "loader": loader_generator.get_state(),
        },
    }

    if improved:
        ft_save(payload, FT_BEST)

    ft_save(payload, FT_LATEST)

    temporary_history = FT_HISTORY.with_suffix(".tmp")
    temporary_history.write_text(json.dumps(ft_history, indent=2))
    os.replace(temporary_history, FT_HISTORY)

    print("\n" + "=" * 72)
    print(f"FINE-TUNING EPOCH {epoch}/{MAX_FINETUNE_EPOCHS}")
    print("=" * 72)
    print(metrics_text("Training", training_metrics))
    print(metrics_text("Validation", validation_metrics))
    print(f"CNN LR: {lrs_used[0]:.2g} → {next_lrs[0]:.2g}")
    print(f"Head LR: {lrs_used[1]:.2g} → {next_lrs[1]:.2g}")
    print(f"Best validation F1: {ft_best_f1:.2%}")
    print(
        "Best model:",
        "starting checkpoint" if ft_best_epoch == 0
        else f"fine-tuning epoch {ft_best_epoch}",
    )
    print(f"No-improvement count: {ft_no_improvement}/{PATIENCE}")
    print(
        "Completed epochs at reduced LR:",
        ft_epochs_at_reduced_lr,
    )
    print("Checkpoints saved to Drive.\n")

    if (
        ft_no_improvement >= PATIENCE
        and ft_epochs_at_reduced_lr >= 2
    ):
        print("Early stopping.")
        break

# ------------------------------------------------------------
# Final summary — no test-set evaluation
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("FINE-TUNING RESULTS")
print("=" * 72)
print("Completed fine-tuning epochs:", ft_history[-1]["epoch"])
print(f"Starting validation F1: {starting_validation['f1']:.2%}")
print(f"Selected validation F1: {ft_best_f1:.2%}")
print(
    "Difference:",
    f"{(ft_best_f1 - starting_validation['f1']) * 100:+.2f}"
    " percentage points",
)

if ft_best_epoch == 0:
    print("No improvement: the starting model remains selected.")
    selected_metrics = starting_validation
else:
    print("Selected fine-tuning epoch:", ft_best_epoch)
    selected_metrics = next(
        record["validation"]
        for record in ft_history
        if record["epoch"] == ft_best_epoch
    )

print(metrics_text("Selected validation", selected_metrics))
print("\nBest checkpoint:", FT_BEST)
print("Latest checkpoint:", FT_LATEST)
print("Test set was not evaluated.")
```

```python
from pathlib import Path
import hashlib
import json
import os
import gc

import numpy as np
import torch
import torchvision
from torchvision.models import (
    mobilenet_v3_small,
    MobileNet_V3_Small_Weights,
)
from torch.utils.data import DataLoader
from tqdm.auto import tqdm

assert "ImprovedVideoDataset" in globals(), (
    "The dataset class is missing. Rerun the earlier data-loader "
    "definition cell first."
)
assert torch.cuda.is_available(), "Enable the GPU first."

DATASET = Path("/content/ViolenceDetectionImproved/safe_dataset")
SOURCE = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/"
    "experiments/improved_v1"
)
FAST_EXPERIMENT = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/"
    "experiments/mobilenet_logistic_v1"
)
CACHE = FAST_EXPERIMENT / "feature_cache"
CACHE.mkdir(parents=True, exist_ok=True)

fast_manifest = json.loads(
    (SOURCE / "split_manifest.json").read_text()
)

# Release previous training objects after stopping their execution.
for name in (
    "model", "optimizer", "scheduler", "checkpoint",
    "trainable_parameters", "finetune_model",
    "finetune_optimizer", "finetune_scheduler",
    "source_checkpoint", "cnn_parameters",
    "head_parameters", "ft_parameters", "payload",
):
    globals().pop(name, None)

gc.collect()
torch.cuda.empty_cache()

DEVICE = torch.device("cuda")

# Use original ImageNet weights, not a fine-tuned checkpoint.
feature_cnn = mobilenet_v3_small(
    weights=MobileNet_V3_Small_Weights.DEFAULT
).features.to(DEVICE).eval()

for parameter in feature_cnn.parameters():
    parameter.requires_grad = False

metadata = {
    "method": "mobilenet_v3_small_imagenet_mean_std_v1",
    "weights": str(MobileNet_V3_Small_Weights.DEFAULT),
    "frames": 16,
    "image_size": 160,
    "sampling": "uniform_endpoints",
    "normalization": "imagenet",
    "augmentation": False,
    "torch_version": torch.__version__,
    "torchvision_version": torchvision.__version__,
    "manifest": fast_manifest,
}
signature = hashlib.sha256(
    json.dumps(metadata, sort_keys=True).encode()
).hexdigest()

metadata_path = FAST_EXPERIMENT / "feature_metadata.json"
if metadata_path.exists():
    assert json.loads(metadata_path.read_text()) == metadata, (
        "Cache settings differ. Use a new experiment folder."
    )
else:
    metadata_path.write_text(json.dumps(metadata, indent=2))

CHUNK_SIZE = 80
fast_features = {}

for split_name in ("train", "validation"):
    records = fast_manifest[split_name]
    feature_parts = []
    label_parts = []

    for start in range(0, len(records), CHUNK_SIZE):
        chunk_records = records[start:start + CHUNK_SIZE]
        cache_file = CACHE / f"{split_name}_{start:04d}.npz"

        if cache_file.exists():
            with np.load(cache_file, allow_pickle=False) as cached:
                assert str(cached["signature"].item()) == signature
                chunk_features = cached["features"].copy()
                chunk_labels = cached["labels"].copy()
            print(f"Reused {cache_file.name}")
        else:
            # Both splits use deterministic, unaugmented frames.
            dataset = ImprovedVideoDataset(
                chunk_records, training=False
            )
            loader = DataLoader(
                dataset,
                batch_size=4,
                shuffle=False,
                num_workers=0,
                pin_memory=True,
            )

            chunk_feature_batches = []
            chunk_label_batches = []

            with torch.inference_mode():
                for clips, labels in tqdm(
                    loader,
                    desc=f"{split_name}: clips {start + 1}–"
                         f"{start + len(chunk_records)}",
                    unit="batch",
                ):
                    clips = clips.to(DEVICE, non_blocking=True)
                    batch, frames, channels, height, width = clips.shape

                    maps = feature_cnn(
                        clips.reshape(
                            batch * frames, channels, height, width
                        )
                    )

                    # 576 visual features per frame.
                    frame_features = maps.mean(dim=(-2, -1))
                    frame_features = frame_features.reshape(
                        batch, frames, -1
                    )

                    # 576 temporal means + 576 temporal standard deviations.
                    summaries = torch.cat(
                        [
                            frame_features.mean(dim=1),
                            frame_features.std(dim=1, unbiased=False),
                        ],
                        dim=1,
                    )

                    chunk_feature_batches.append(
                        summaries.cpu().numpy()
                    )
                    chunk_label_batches.append(
                        labels.numpy().astype(np.int64)
                    )

            chunk_features = np.concatenate(chunk_feature_batches)
            chunk_labels = np.concatenate(chunk_label_batches)

            assert np.isfinite(chunk_features).all()

            temporary = cache_file.with_suffix(".tmp")
            with temporary.open("wb") as file:
                np.savez_compressed(
                    file,
                    features=chunk_features,
                    labels=chunk_labels,
                    signature=np.array(signature),
                )
            os.replace(temporary, cache_file)

        assert chunk_features.shape == (len(chunk_records), 1152)
        assert np.array_equal(
            chunk_labels,
            np.array([record["label"] for record in chunk_records]),
        )

        feature_parts.append(chunk_features)
        label_parts.append(chunk_labels)

    fast_features[split_name] = (
        np.concatenate(feature_parts),
        np.concatenate(label_parts),
    )

X_train, y_train = fast_features["train"]
X_validation, y_validation = fast_features["validation"]

print("\nFeature extraction complete.")
print("Training features:", X_train.shape)
print("Validation features:", X_validation.shape)
print("Cache location:", CACHE)
print("Test videos were not processed.")
```

```python
import warnings
import joblib

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    log_loss,
)
from sklearn.exceptions import ConvergenceWarning

def classification_metrics(y_true, probabilities):
    predictions = (probabilities >= 0.5).astype(int)
    return {
        "loss": float(log_loss(
            y_true, probabilities, labels=[0, 1]
        )),
        "accuracy": float(accuracy_score(y_true, predictions)),
        "precision": float(precision_score(
            y_true, predictions, zero_division=0
        )),
        "recall": float(recall_score(
            y_true, predictions, zero_division=0
        )),
        "f1": float(f1_score(
            y_true, predictions, zero_division=0
        )),
    }

best_pipeline = None
best_result = None
results = []

# Smaller C means stronger regularisation.
for C in (0.0001, 0.001, 0.01, 0.1, 1.0):
    pipeline = Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            C=C,
            solver="lbfgs",
            max_iter=5000,
            random_state=42,
        )),
    ])

    # Surface convergence failures instead of silently accepting them.
    with warnings.catch_warnings():
        warnings.simplefilter("error", ConvergenceWarning)
        pipeline.fit(X_train, y_train)

    training = classification_metrics(
        y_train, pipeline.predict_proba(X_train)[:, 1]
    )
    validation = classification_metrics(
        y_validation,
        pipeline.predict_proba(X_validation)[:, 1],
    )

    result = {
        "C": C,
        "training": training,
        "validation": validation,
    }
    results.append(result)

    print(f"\nC = {C}")
    for name, metrics in (
        ("Training", training),
        ("Validation", validation),
    ):
        print(
            f"{name}: loss {metrics['loss']:.4f}, "
            f"accuracy {metrics['accuracy']:.1%}, "
            f"precision {metrics['precision']:.1%}, "
            f"recall {metrics['recall']:.1%}, "
            f"F1 {metrics['f1']:.1%}"
        )

    # On equal F1, keep the earlier, more regularised model.
    if (
        best_result is None
        or validation["f1"] > best_result["validation"]["f1"]
    ):
        best_result = result
        best_pipeline = pipeline

joblib.dump(
    best_pipeline,
    FAST_EXPERIMENT / "best_logistic_classifier.joblib",
)

(FAST_EXPERIMENT / "classifier_results.json").write_text(
    json.dumps(
        {"selected": best_result, "all_results": results},
        indent=2,
    )
)

print("\n" + "=" * 60)
print("SELECTED LOGISTIC-REGRESSION MODEL")
print("=" * 60)
print("C:", best_result["C"])

for key, value in best_result["validation"].items():
    if key == "loss":
        print(f"Validation loss: {value:.4f}")
    else:
        print(f"Validation {key}: {value:.2%}")

print("\nPrevious LSTM-32 validation F1: 67.04%")
print("Saved in:", FAST_EXPERIMENT)
print("Test set was not evaluated.")
```

```python
from pathlib import Path

KNN_ROOT = Path(
    "/content/google_drive/MyDrive/ViolenceDetection/KNN_Experiment"
)
SCRIPT = KNN_ROOT / "scripts/train_knn.py"

script_code = r'''
from pathlib import Path
import hashlib
import json
import platform

import joblib
import numpy as np
import sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, log_loss, confusion_matrix,
)
from tqdm.auto import tqdm

ROOT = Path(__file__).resolve().parents[1]
MODEL_PATH = ROOT / "models/best_knn.joblib"
RESULTS_PATH = ROOT / "results/validation_results.json"

assert not MODEL_PATH.exists() and not RESULTS_PATH.exists(), (
    "KNN results already exist. Read the saved results before rerunning."
)

metadata = json.loads((ROOT / "feature_metadata.json").read_text())
signature = hashlib.sha256(
    json.dumps(metadata, sort_keys=True).encode()
).hexdigest()


def load_features(split):
    records = metadata["manifest"][split]
    expected_chunks = (len(records) + 79) // 80
    files = sorted((ROOT / "features").glob(f"{split}_*.npz"))
    assert len(files) == expected_chunks

    feature_parts, label_parts = [], []

    for start in range(0, len(records), 80):
        path = ROOT / "features" / f"{split}_{start:04d}.npz"
        with np.load(path, allow_pickle=False) as cache:
            assert str(cache["signature"].item()) == signature
            features = cache["features"].copy()
            labels = cache["labels"].copy()

        expected_labels = np.array([
            record["label"] for record in records[start:start + 80]
        ])
        assert features.shape == (len(expected_labels), 1152)
        assert np.isfinite(features).all()
        assert np.array_equal(labels, expected_labels)

        feature_parts.append(features)
        label_parts.append(labels)

    return np.concatenate(feature_parts), np.concatenate(label_parts)


def metrics(y, probabilities):
    predictions = (probabilities >= 0.5).astype(int)
    return {
        "loss": float(log_loss(y, probabilities, labels=[0, 1])),
        "accuracy": float(accuracy_score(y, predictions)),
        "precision": float(precision_score(
            y, predictions, zero_division=0
        )),
        "recall": float(recall_score(
            y, predictions, zero_division=0
        )),
        "f1": float(f1_score(y, predictions, zero_division=0)),
    }


X_train, y_train = load_features("train")
X_validation, y_validation = load_features("validation")

print("Training features:", X_train.shape)
print("Validation features:", X_validation.shape)

# Fit scaling ONLY on training features.
scaler = StandardScaler()
train_scaled = scaler.fit_transform(X_train)
validation_scaled = scaler.transform(X_validation)

results = []
best_f1 = -1.0
best_pipeline = None
best_result = None

# Three PCA sizes × four neighbour counts × two voting schemes.
progress = tqdm(total=24, desc="KNN configurations")

for components in (32, 64, 128):
    # Fit PCA ONLY on training features.
    pca = PCA(
        n_components=components,
        svd_solver="randomized",
        random_state=42,
    )
    train_reduced = pca.fit_transform(train_scaled)
    validation_reduced = pca.transform(validation_scaled)

    for k in (3, 5, 11, 21):
        for weights in ("uniform", "distance"):
            knn = KNeighborsClassifier(
                n_neighbors=k,
                weights=weights,
                metric="euclidean",
                algorithm="brute",
                n_jobs=-1,
            )
            knn.fit(train_reduced, y_train)

            probabilities = knn.predict_proba(
                validation_reduced
            )[:, 1]
            validation = metrics(y_validation, probabilities)

            result = {
                "pca_components": components,
                "neighbors": k,
                "weights": weights,
                "explained_variance": float(
                    pca.explained_variance_ratio_.sum()
                ),
                "validation": validation,
            }
            results.append(result)

            # Predeclared tie-break: keep the first equal-F1 candidate.
            if validation["f1"] > best_f1:
                best_f1 = validation["f1"]
                best_result = result
                best_pipeline = Pipeline([
                    ("scaler", scaler),
                    ("pca", pca),
                    ("classifier", knn),
                ])

            progress.update(1)
            progress.set_postfix(best_f1=f"{best_f1:.2%}")

progress.close()

probabilities = best_pipeline.predict_proba(X_validation)[:, 1]
predictions = (probabilities >= 0.5).astype(int)

# Verify the assembled pipeline reproduces the selected result.
assert abs(metrics(y_validation, probabilities)["f1"] - best_f1) < 1e-12

joblib.dump(best_pipeline, MODEL_PATH)

report = {
    "selected": best_result,
    "all_configurations": results,
    "selection_metric": "validation F1",
    "classification_threshold": 0.5,
    "tie_break": "first configuration with equal F1",
    "reference_logistic_validation_f1": 0.757396449704142,
    "feature_signature": signature,
    "confusion_matrix": confusion_matrix(
        y_validation, predictions, labels=[0, 1]
    ).tolist(),
    "test_evaluated": False,
}

RESULTS_PATH.write_text(json.dumps(report, indent=2))

np.savez_compressed(
    ROOT / "results/validation_predictions.npz",
    paths=np.array([
        record["path"]
        for record in metadata["manifest"]["validation"]
    ]),
    labels=y_validation,
    fight_probabilities=probabilities,
    predictions=predictions,
)

(ROOT / "results/package_versions.json").write_text(json.dumps({
    "python": platform.python_version(),
    "numpy": np.__version__,
    "sklearn": sklearn.__version__,
    "joblib": joblib.__version__,
}, indent=2))

lines = [
    "# KNN validation results",
    "",
    f"PCA components: {best_result['pca_components']}",
    f"Neighbours: {best_result['neighbors']}",
    f"Voting weights: {best_result['weights']}",
    "Classification threshold: 0.5",
    "",
]
for name, value in best_result["validation"].items():
    lines.append(
        f"{name}: {value:.4f}" if name == "loss"
        else f"{name}: {value:.2%}"
    )
lines.extend([
    "",
    "Selected from 24 configurations using validation F1.",
    "Scaling and PCA were fitted on training data only.",
    "Test set was not evaluated.",
])
(ROOT / "results/KNN_SUMMARY.md").write_text("\n".join(lines))

print("\n" + "=" * 60)
print("BEST KNN VALIDATION RESULTS")
print("=" * 60)
print("PCA components:", best_result["pca_components"])
print("Neighbours:", best_result["neighbors"])
print("Voting:", best_result["weights"])

for name, value in best_result["validation"].items():
    if name == "loss":
        print(f"Validation {name}: {value:.4f}")
    else:
        print(f"Validation {name}: {value:.2%}")

print("\nLogistic regression validation F1: 75.74%")
print(f"KNN validation F1: {best_f1:.2%}")
print("\nModel saved:", MODEL_PATH)
print("Results saved:", RESULTS_PATH)
print("Test set was not evaluated.")
'''

assert not (KNN_ROOT / "models/best_knn.joblib").exists(), (
    "A saved KNN model already exists. Stop before rerunning."
)

SCRIPT.write_text(script_code, encoding="utf-8")
print("Script saved:", SCRIPT)

# Run the saved script in this Colab kernel.
exec(compile(script_code, str(SCRIPT), "exec"), {
    "__name__": "__main__",
    "__file__": str(SCRIPT),
})
```

```python
from pathlib import Path
import json
import zipfile
from google.colab import files

ROOT = Path("/content/google_drive/MyDrive/ViolenceDetection")
DATASET = ROOT / "RWF-2000/safe_dataset"

metadata = json.loads(
    (ROOT / "experiments/mobilenet_logistic_v1/"
     "feature_metadata.json").read_text()
)

previous_clips = {
    "train/NonFight/0f76804c04869b44722c.avi",
    "train/Fight/8c6bc3fda59678f2292f.avi",
}

archive = Path("/content/more_validation_videos.zip")
sources = {}

with zipfile.ZipFile(archive, "w", zipfile.ZIP_DEFLATED) as bundle:
    for label, name in [(0, "NonFight"), (1, "Fight")]:
        selected = [
            record
            for record in metadata["manifest"]["validation"]
            if record["label"] == label
            and record["path"] not in previous_clips
        ][:2]

        assert len(selected) == 2

        for number, record in enumerate(selected, start=2):
            source = DATASET / record["path"]
            assert source.is_file(), f"Missing: {source}"

            filename = f"{name}_{number}{source.suffix}"
            bundle.write(source, filename)
            sources[filename] = record["path"]
            print(f"Added {filename}: {record['path']}")

    bundle.writestr("clip_sources.json", json.dumps(sources, indent=2))

files.download(str(archive))
```
